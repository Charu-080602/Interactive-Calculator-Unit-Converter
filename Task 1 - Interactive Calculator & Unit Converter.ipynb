```python
def basic_calculator():
    print("\n--- Basic Calculator ---")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")

    while True:
        choice = input("Select operation (1-4): ").strip()
        if choice in ['1', '2', '3', '4']:
            break
        print("Invalid choice! Please enter a number from 1 to 4.")

    while True:
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            break
        except ValueError:
            print("Invalid input! Please enter valid numeric values.")

    if choice == '1':
        print(f"Result: {num1} + {num2} = {num1 + num2}")
    elif choice == '2':
        print(f"Result: {num1} - {num2} = {num1 - num2}")
    elif choice == '3':
        print(f"Result: {num1} * {num2} = {num1 * num2}")
    elif choice == '4':
        if num2 == 0:
            print("Error: Division by zero is not allowed.")
        else:
            print(f"Result: {num1} / {num2} = {num1 / num2}")


def unit_converter():
    print("\n--- Unit Converter ---")
    print("1. Kilometers to Miles")
    print("2. Miles to Kilometers")
    print("3. Celsius to Fahrenheit")
    print("4. Fahrenheit to Celsius")

    while True:
        choice = input("Select conversion type (1-4): ").strip()
        if choice in ['1', '2', '3', '4']:
            break
        print("Invalid choice! Please enter a number from 1 to 4.")

    while True:
        try:
            val = float(input("Enter value to convert: "))
            break
        except ValueError:
            print("Invalid input! Please enter a numeric value.")

    if choice == '1':
        print(f"Result: {val} km = {val * 0.621371:.2f} miles")
    elif choice == '2':
        print(f"Result: {val} miles = {val / 0.621371:.2f} km")
    elif choice == '3':
        print(f"Result: {val}°C = {(val * 9/5) + 32:.2f}°F")
    elif choice == '4':
        print(f"Result: {val}°F = {(val - 32) * 5/9:.2f}°C")


def currency_converter():
    print("\n--- Currency Converter (USD / INR / EUR) ---")
    print("1. USD to INR")
    print("2. INR to USD")
    print("3. EUR to INR")
    print("4. INR to EUR")

    # Fixed exchange rates (for demo CLI program)
    usd_to_inr = 83.5
    eur_to_inr = 90.0

    while True:
        choice = input("Select currency pair (1-4): ").strip()
        if choice in ['1', '2', '3', '4']:
            break
        print("Invalid choice! Please enter a number from 1 to 4.")

    while True:
        try:
            amount = float(input("Enter amount: "))
            if amount < 0:
                print("Amount cannot be negative.")
                continue
            break
        except ValueError:
            print("Invalid input! Please enter a numeric amount.")

    if choice == '1':
        print(f"Result: ${amount} USD = ₹{amount * usd_to_inr:.2f} INR")
    elif choice == '2':
        print(f"Result: ₹{amount} INR = ${amount / usd_to_inr:.2f} USD")
    elif choice == '3':
        print(f"Result: €{amount} EUR = ₹{amount * eur_to_inr:.2f} INR")
    elif choice == '4':
        print(f"Result: ₹{amount} INR = €{amount / eur_to_inr:.2f} EUR")


def main():
    while True:
        print("\n=========================================")
        print(" INTERACTIVE CALCULATOR & CONVERTER CLI ")
        print("=========================================")
        print("1. Basic Arithmetic")
        print("2. Unit Conversion")
        print("3. Currency Conversion")
        print("4. Exit")

        choice = input("Choose an option (1-4): ").strip()

        if choice == '1':
            basic_calculator()
        elif choice == '2':
            unit_converter()
        elif choice == '3':
            currency_converter()
        elif choice == '4':
            print("\nThank you for using the CLI tool! Goodbye.")
            break
        else:
            print("Invalid selection! Please enter 1, 2, 3, or 4.")


if __name__ == "__main__":
    main()
```
