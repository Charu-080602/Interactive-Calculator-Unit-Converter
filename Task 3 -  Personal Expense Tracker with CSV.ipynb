```python
import csv
import os

FILENAME = "expenses.csv"


def initialize_file():
    if not os.path.exists(FILENAME):
        with open(FILENAME, mode="w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["Date", "Category", "Description", "Amount"])


def add_expense():
    print("\n--- Add New Expense ---")
    date = input("Enter date (YYYY-MM-DD): ").strip()
    category = input("Enter category (e.g., Food, Travel): ").strip()
    description = input("Enter description: ").strip()

    while True:
        try:
            amount = float(input("Enter amount: "))
            if amount < 0:
                print("Amount cannot be negative.")
                continue
            break
        except ValueError:
            print("Invalid input! Please enter a valid number.")

    with open(FILENAME, mode="a", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow([date, category, description, amount])
    print("Expense added successfully!")


def view_expenses():
    print("\n--- All Expenses ---")
    initialize_file()
    with open(FILENAME, mode="r", encoding="utf-8") as file:
        reader = csv.reader(file)
        rows = list(reader)
        if len(rows) <= 1:
            print("No expenses recorded yet.")
            return
        for row in rows:
            print(f"{row[0]} | {row[1]} | {row[2]} | ₹{row[3]}")


def filter_expenses():
    print("\n--- Filter Expenses by Category ---")
    initialize_file()
    target_category = input("Enter category to filter by: ").strip().lower()

    with open(FILENAME, mode="r", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        found = False
        print(f"\nExpenses in category '{target_category}':")
        print("-" * 40)
        for row in reader:
            if row["Category"].lower() == target_category:
                print(f"{row['Date']} | {row['Description']} | ₹{row['Amount']}")
                found = True
        if not found:
            print("No expenses found for this category.")


def expense_summary():
    print("\n--- Category Summary Statistics ---")
    initialize_file()
    category_totals = {}
    total_spent = 0.0

    with open(FILENAME, mode="r", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            cat = row["Category"]
            amt = float(row["Amount"])
            category_totals[cat] = category_totals.get(cat, 0.0) + amt
            total_spent += amt

    if total_spent == 0:
        print("No expenses to summarize.")
        return

    print(f"Total Spending: ₹{total_spent:.2f}\n")
    print("Category Breakdown:")
    for cat, amt in category_totals.items():
        print(f" - {cat}: ₹{amt:.2f}")


def main():
    initialize_file()
    while True:
        print("\n=========================================")
        print(" PERSONAL EXPENSE TRACKER ")
        print("=========================================")
        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. Filter Expenses by Category")
        print("4. View Category Summary Statistics")
        print("5. Exit")

        choice = input("Choose an option (1-5): ").strip()
        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            filter_expenses()
        elif choice == "4":
            expense_summary()
        elif choice == "5":
            print("Exiting Expense Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-5.")


if __name__ == "__main__":
    main()
```
