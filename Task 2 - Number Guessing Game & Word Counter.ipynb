```python
import random
import re
from collections import Counter


def number_guessing_game():
    print("\n--- Number Guessing Game ---")
    secret_number = random.randint(1, 100)
    attempts = 0
    max_attempts = 10

    print("I have chosen a number between 1 and 100.")
    print(f"You have {max_attempts} attempts to guess it correctly!\n")

    while attempts < max_attempts:
        try:
            guess = int(
                input(
                    f"Attempt {attempts + 1}/{max_attempts} - Enter your guess: "
                )
            )
        except ValueError:
            print("Invalid input! Please enter a valid integer.")
            continue

        attempts += 1

        if guess < secret_number:
            print("Too low! Try a higher number.")
        elif guess > secret_number:
            print("Too high! Try a lower number.")
        else:
            score = (max_attempts - attempts + 1) * 10
            print(
                f"\nCongratulations! You guessed the correct number {secret_number} in {attempts} attempt(s)."
            )
            print(f"Your Score: {score} / 100")
            return

    print(
        f"\nGame Over! You ran out of attempts. The secret number was {secret_number}."
    )


def word_counter_tool():
    print("\n--- File-based Word Counter & Frequency Analysis ---")
    file_option = (
        input(
            "Do you want to (1) Enter text directly or (2) Read from a file path? [1/2]: "
        )
        .strip()
    )

    text_content = ""

    if file_option == "2":
        file_path = input("Enter the file path (e.g., sample.txt): ").strip()
        try:
            with open(file_path, "r", encoding="utf-8") as file:
                text_content = file.read()
            print(f"Successfully loaded content from '{file_path}'.")
        except FileNotFoundError:
            print(
                f"Error: File '{file_path}' not found. Switching to direct text input."
            )
            text_content = input("Enter your text paragraph: ")
        except Exception as e:
            print(f"An error occurred: {e}")
            return
    else:
        text_content = input("Enter or paste your text paragraph: ")

    if not text_content.strip():
        print("No text provided. Exiting word counter.")
        return

    words = re.findall(r"\b\w+\b", text_content.lower())
    total_words = len(words)
    unique_words = len(set(words))
    word_counts = Counter(words)

    print("\n" + "=" * 40)
    print(" WORD ANALYSIS RESULTS ")
    print("=" * 40)
    print(f"Total Word Count:  {total_words}")
    print(f"Unique Word Count: {unique_words}")

    print("\nTop 5 Most Frequent Words:")
    print("-" * 30)
    for word, count in word_counts.most_common(5):
        print(f"'{word}': {count} time(s)")
    print("=" * 40)


def main():
    while True:
        print("\n=========================================")
        print(" NUMBER GUESSER & WORD COUNTER TOOL ")
        print("=========================================")
        print("1. Play Number Guessing Game")
        print("2. Run Word Counter & Frequency Analysis")
        print("3. Exit")

        choice = input("Choose an option (1-3): ").strip()

        if choice == "1":
            number_guessing_game()
        elif choice == "2":
            word_counter_tool()
        elif choice == "3":
            print("\nThank you for using the tool! Goodbye.")
            break
        else:
            print("Invalid selection! Please enter 1, 2, or 3.")


if __name__ == "__main__":
    main()
```
